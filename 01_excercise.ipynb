```python
from google.cloud import bigquery
import pandas as pd

print("Libraries imported successfully!")
```

```python
client = bigquery.Client(project="introtosql-513910")

print("BigQuery connection successful!")
print("Project:", client.project)
```

```python
dataset_ref = client.dataset("chicago_crime", project="bigquery-public-data")
dataset = client.get_dataset(dataset_ref)

# Pobieramy listę tabel
tables = list(client.list_tables(dataset))

# Liczymy tabele
num_tables = len(tables)

print(num_tables)
```

num_tables = len(list(client.list_tables(dataset)))

Use the list_tables() method to get a list of the tables in the dataset.

```python
table_ref = dataset_ref.table("crime")
table = client.get_table(table_ref)
num_timestamp_fields = 0

for column in table.schema:
    if column.field_type == "TIMESTAMP":
        num_timestamp_fields += 1

print(num_timestamp_fields)
```
